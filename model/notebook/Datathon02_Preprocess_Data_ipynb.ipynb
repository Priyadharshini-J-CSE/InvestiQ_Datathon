```python
import pandas as pd

BASE_PATH = "/content/drive/MyDrive/database"

fir = pd.read_excel(f"{BASE_PATH}/fir.xlsx")
persons = pd.read_excel(f"{BASE_PATH}/persons.xlsx")
criminals = pd.read_excel(f"{BASE_PATH}/criminals.xlsx")
wanted = pd.read_excel(f"{BASE_PATH}/wanted.xlsx")
cases = pd.read_excel(f"{BASE_PATH}/cases.xlsx")
charges = pd.read_excel(f"{BASE_PATH}/charges.xlsx")
arrests = pd.read_excel(f"{BASE_PATH}/arrests.xlsx")
convictions = pd.read_excel(f"{BASE_PATH}/convictions.xlsx")
warrants = pd.read_excel(f"{BASE_PATH}/warrants.xlsx")
officers = pd.read_excel(f"{BASE_PATH}/officers.xlsx")
evidence = pd.read_excel(f"{BASE_PATH}/evidence.xlsx")
users = pd.read_excel(f"{BASE_PATH}/users.xlsx")

print("All Excel files loaded successfully!")
```

```python
datasets = {
    "FIR": fir,
    "Persons": persons,
    "Criminals": criminals,
    "Wanted": wanted,
    "Cases": cases,
    "Charges": charges,
    "Arrests": arrests,
    "Convictions": convictions,
    "Warrants": warrants,
    "Officers": officers,
    "Evidence": evidence,
    "Users": users
}

for name, df in datasets.items():
    print(f"{name}: {df.shape}")
```

```python
for name, df in datasets.items():
    df.fillna("Unknown", inplace=True)

print("Missing Values Filled")
```

```python
merged = fir.merge(

    criminals,

    on="Criminal_ID",

    how="left"

)
```

```python
merged = merged.merge(

    persons,

    on="Person_ID",

    how="left"

)
```

```python
merged = merged.merge(

    officers,

    on="Officer_ID",

    how="left"

)
```

```python
merged = merged.merge(

    cases,

    on="FIR_ID",

    how="left"

)
```

```python
merged = merged.merge(

    charges,

    on=["FIR_ID","Criminal_ID"],

    how="left"

)
```

```python
merged = merged.merge(

    arrests,

    on=["FIR_ID","Criminal_ID"],

    how="left"

)
```

```python
merged = merged.merge(

    convictions,

    on="Criminal_ID",

    how="left"

)
```

```python
merged = merged.merge(

    warrants,

    on="Criminal_ID",

    how="left"

)
```

```python
merged = merged.merge(

    wanted,

    on="Criminal_ID",

    how="left"

)
```

```python
merged = merged.merge(

    evidence,

    on="FIR_ID",

    how="left"

)
```

```python
merged = merged.loc[:, ~merged.columns.duplicated()]
```

```python
print(merged.shape)

merged.head()
```

```python
date_columns = merged.select_dtypes(include=["datetime64"]).columns

for col in date_columns:

    merged[col] = merged[col].astype(str)
```

```python
for col in merged.columns:

    if merged[col].dtype == object:

        merged[col] = merged[col].astype(str).str.strip()
```

```python
os.makedirs("processed", exist_ok=True)

merged.to_excel(

    "processed/merged_police_data.xlsx",

    index=False

)

print("Merged Dataset Saved Successfully")
```
