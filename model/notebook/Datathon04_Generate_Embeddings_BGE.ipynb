```python
# Uncomment if your files are in Google Drive
# from google.colab import drive
# drive.mount('/content/drive')

import os
import pickle  # Fixed: Added missing import
import pandas as pd

# Verify drive contents if mounted
# print(os.listdir("/content/drive/MyDrive"))

BASE_PATH = "/content/drive/MyDrive/database"
file_path = f"{BASE_PATH}/documents.pkl"

# Fixed: Open the file in 'rb' (read binary) mode before loading
with open(file_path, "rb") as f:
    documents = pickle.load(f)

print("Total Documents:", len(documents))
```

```python
import pickle
import numpy as np

from tqdm import tqdm

from sentence_transformers import SentenceTransformer
```

```python
model = SentenceTransformer(
    "BAAI/bge-small-en-v1.5"
)

print("Embedding Model Loaded")
```

```python
texts = []

metadata = []

for doc in documents:

    texts.append(doc.page_content)

    metadata.append(doc.metadata)

print("Texts :", len(texts))

print("Metadata :", len(metadata))
```

```python
embeddings = model.encode(

    texts,

    show_progress_bar=True,

    convert_to_numpy=True,

    normalize_embeddings=True

)
```

```python
print(embeddings.shape)
```

```python
np.save("embeddings.npy", embeddings)

print("Embeddings Saved")
```

```python
import pickle

with open("metadata.pkl", "wb") as f:
    pickle.dump(metadata, f)

print("Metadata Saved")
```
