```python
import pandas as pd

BASE_PATH = "/content/drive/MyDrive/database"

df = pd.read_excel(f"{BASE_PATH}/merged_police_data.xlsx")


print(df.shape)

df.head()
```

```python
df.fillna("Not Available", inplace=True)

print("Missing Values Filled")
```

```python
print(df.columns.tolist())
```

```python
documents = []

for _, row in df.iterrows():

    text = f"""

================ FIR DETAILS ================

FIR Number : {row['FIR_Number']}

Case Number : {row['Case_Number']}

Crime Category : {row['Crime_Category']}

Crime Type : {row['Crime_Type']}

IPC Sections : {row['IPC_Sections']}

Description :

{row['Description_x']}

Investigation Status :

{row['Investigation_Status']}

Charge Sheet :

{row['Chargesheet_Status']}

Police Station :

{row['Police_Station']}

District :

{row['District_x']}

Incident Date :

{row['Incident_Date']}

Incident Time :

{row['Incident_Time']}


================ PERSON DETAILS ================

Person Name :

{row['Full_Name']}

Gender :

{row['Gender']}

DOB :

{row['DOB']}

Occupation :

{row['Occupation']}

Address :

{row['Address']}

Mobile :

{row['Mobile_x']}


================ CRIMINAL DETAILS ================

Alias :

{row['Alias']}

Gang :

{row['Gang_Name']}

Risk Level :

{row['Risk_Level']}

Previous Cases :

{row['Previous_Cases']}

Arrest Count :

{row['Arrest_Count']}

Repeat Offender :

{row['Repeat_Offender']}

Bail Status :

{row['Bail_Status']}

Current Status :

{row['Current_Status']}

Wanted :

{row['Wanted_Status']}


================ OFFICER DETAILS ================

Officer Name :

{row['Name']}

Rank :

{row['Rank']}

Station :

{row['Station']}

Officer District :

{row['District']}


================ CASE DETAILS ================

Court :

{row['Court_Name']}

Judge :

{row['Judge']}

Status :

{row['Status_x']}

Verdict :

{row['Verdict']}


================ CHARGES ================

IPC :

{row['IPC']}

Section :

{row['Section_Name']}

Severity :

{row['Severity']}


================ ARREST DETAILS ================

Arrest Date :

{row['Arrest_Date']}

Location :

{row['Arrest_Location']}

Arresting Officer :

{row['Arresting_Officer']}


================ CONVICTION ================

Sentence :

{row['Sentence']}

Fine :

{row['Fine']}

Prison Term :

{row['Prison_Term']}

Verdict Date :

{row['Verdict_Date']}


================ WARRANT ================

Type :

{row['Type_x']}

Issue Date :

{row['Issue_Date']}

Expiry Date :

{row['Expiry_Date']}

Status :

{row['Status_y']}


================ WANTED ================

Reward :

{row['Reward']}

Reason :

{row['Reason']}

Last Seen :

{row['Last_Seen_Date']}

Location :

{row['Last_Location']}

Issued By :

{row['Issued_By']}

Active :

{row['Active']}


================ EVIDENCE ================

Evidence Type :

{row['Type_y']}

Evidence Description :

{row['Description_y']}

File :

{row['File_Name']}

Uploaded By :

{row['Uploaded_By']}

"""

    metadata = {

        "fir_number": row["FIR_Number"],

        "case_number": row["Case_Number"],

        "crime": row["Crime_Type"],

        "district": row["District_x"],

        "station": row["Police_Station"],

        "criminal": row["Full_Name"],

        "officer": row["Name"],

        "rank": row["Rank"],

        "status": row["Investigation_Status"]

    }

    documents.append(

        Document(

            page_content=text,

            metadata=metadata

        )

    )

print("Documents Created :", len(documents))
```
